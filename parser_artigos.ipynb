```python
import json
from pathlib import Path
from lib.parser.container import parser_container
from lib.parser.publication import parser_publication
from lib.parser.contributors import parser_contributor
from lib.parser.funders import parse_funder
from lib.parser.references import parser_references
```

```python
path_artigos = Path("data/artigos/val.jsonl")
artigos = []
with open(path_artigos, 'r', encoding='utf-8') as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        artigos.append(json.loads(line))
len(artigos)
```

```python
artigos_normalizados = []
for artigo in artigos:
    container = parser_container(artigo)
    publication = parser_publication(artigo)
    contributors = parser_contributor(artigo)
    funders = parse_funder(artigo)
    references = parser_references(artigo)
    print(publication['title'])
    norm = {
        "container": container,
        "publication": publication,
        "contributors": contributors,
        "funders": funders,
        "references": references
    }
    artigos_normalizados.append(norm)
```

```python
len(artigos_normalizados)
```

```python
with open("data/artigos/artigos_normalizados.jsonl", "w", encoding="utf-8") as f:
    for artigo in artigos_normalizados:
        json.dump(artigo, f, ensure_ascii=False, default=str)
        f.write("\n")
```
